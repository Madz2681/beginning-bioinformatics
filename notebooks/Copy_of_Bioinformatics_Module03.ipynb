### Problem Description: Square of the Hypotenuse

**Given:** Two positive integers `a` and `b`, each less than 1000.

**Task:** Return the integer corresponding to the square of the hypotenuse of the right triangle whose legs have lengths `a` and `b`.

This can be solved using the Pythagorean theorem, which states that for a right triangle with legs `a` and `b` and hypotenuse `c`, $a^2 + b^2 = c^2$. We need to calculate $c^2$.

```python
# It's generally good practice to allow for variable inputs, rather than hardcoding.
# For demonstration, we'll use the values a=807 and b=825 as in the previous example.
# If you wanted to take user input, you could use:
# a = int(input("Enter the value for leg 'a': "))
# b = int(input("Enter the value for leg 'b': "))

a = 3
b = 5

# Calculate the square of the hypotenuse
square_of_hypotenuse = a**2 + b**2

# Print the result
print(f"The square of the hypotenuse for legs a={a} and b={b} is: {square_of_hypotenuse}")
```

```python
# Rosalind 8
with open("rosalind_prot.txt", "r") as data:
    rna = data.readline().strip()

codon_table = {
    "UUU":"F", "UUC":"F", "UUA":"L", "UUG":"L",
    "UCU":"S", "UCC":"S", "UCA":"S", "UCG":"S",
    "UAU":"Y", "UAC":"Y", "UAA":"Stop", "UAG":"Stop",
    "UGU":"C", "UGC":"C", "UGA":"Stop", "UGG":"W",

    "CUU":"L", "CUC":"L", "CUA":"L", "CUG":"L",
    "CCU":"P", "CCC":"P", "CCA":"P", "CCG":"P",
    "CAU":"H", "CAC":"H", "CAA":"Q", "CAG":"Q",
    "CGU":"R", "CGC":"R", "CGA":"R", "CGG":"R",

    "AUU":"I", "AUC":"I", "AUA":"I", "AUG":"M",
    "ACU":"T", "ACC":"T", "ACA":"T", "ACG":"T",
    "AAU":"N", "AAC":"N", "AAA":"K", "AAG":"K",
    "AGU":"S", "AGC":"S", "AGA":"R", "AGG":"R",

    "GUU":"V", "GUC":"V", "GUA":"V", "GUG":"V",
    "GCU":"A", "GCC":"A", "GCA":"A", "GCG":"A",
    "GAU":"D", "GAC":"D", "GAA":"E", "GAG":"E",
    "GGU":"G", "GGC":"G", "GGA":"G", "GGG":"G"
}

protein = ""

for i in range(0, len(rna), 3):
    codon = rna[i:i+3]

    if codon_table[codon] == "Stop":
        break

    protein += codon_table[codon]

print(protein)
```

```python
# rosalind 9

from Bio import SeqIO

records = SeqIO.parse("rosalind_gc.txt", "fasta")

max_gc = 0
max_id = ""

for record in records:
    gc = (record.seq.count("G") + record.seq.count("C")) / len(record.seq) * 100

    if gc > max_gc:
        max_gc = gc
        max_id = record.id

print(max_id)
print(max_gc)
```
